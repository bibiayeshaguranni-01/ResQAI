```python
def standardize_columns(df):
    df = df.copy()
    df.columns = df.columns.str.strip()

    rename_map = {
        "Description": "description",
        "Alertlevel": "alertlevel",
        "Alertscore": "alertscore",
        "Episodealertlevel": "episodealertlevel",
        "Episodealertscore": "episodealertscore",
        "Country": "country",
        "Fromdate": "fromdate",
        "Todate": "todate",
        "Iso3": "iso3",
        "Gdacs id": "gdacs_id",
        "GDACS ID": "gdacs_id",
        "GDACS_ID": "gdacs_id",
        "Longitude": "longitude",
        "Latitude": "latitude"
    }

    df = df.rename(columns=rename_map)

    return df
```

```python
drought = standardize_columns(drought)
earthquake = standardize_columns(earthquake)
eruption = standardize_columns(eruption)
flood = standardize_columns(flood)
forest_fires = standardize_columns(forest_fires)
tropical_cyclone = standardize_columns(tropical_cyclone)
```

```python
# Make copies so the original raw datasets remain unchanged
drought_ml = drought_raw.copy()
earthquake_ml = earthquake_raw.copy()
eruption_ml = eruption_raw.copy()
flood_ml = flood_raw.copy()
forest_fires_ml = forest_fires_raw.copy()
tropical_cyclone_ml = tropical_cyclone_raw.copy()


# Standardize column names
def standardize_ml_columns(df):
    df = df.copy()

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("(", "", regex=False)
        .str.replace(")", "", regex=False)
    )

    return df


drought_ml = standardize_ml_columns(drought_ml)
earthquake_ml = standardize_ml_columns(earthquake_ml)
eruption_ml = standardize_ml_columns(eruption_ml)
flood_ml = standardize_ml_columns(flood_ml)
forest_fires_ml = standardize_ml_columns(forest_fires_ml)
tropical_cyclone_ml = standardize_ml_columns(tropical_cyclone_ml)


# Add disaster type
drought_ml["disaster_type"] = "Drought"
earthquake_ml["disaster_type"] = "Earthquake"
eruption_ml["disaster_type"] = "Eruption"
flood_ml["disaster_type"] = "Flood"
forest_fires_ml["disaster_type"] = "Forest Fire"
tropical_cyclone_ml["disaster_type"] = "Tropical Cyclone"


print("Column standardization completed.")
```

```python
from sklearn.model_selection import train_test_split

feature_columns = [
    "alertlevel",
    "alertscore",
    "episodealertscore",
    "longitude",
    "latitude",
    "year",
    "month",
    "duration_days_calculated",
]

X = rich_disaster_dataset[feature_columns].copy()
y = rich_disaster_dataset["disaster_type"].copy()

# Confirm the target is valid
assert X.shape[0] == y.shape[0]
assert y.notna().all()
assert "disaster_type" not in X.columns

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Missing values in X:")
print(X.isnull().sum())
print("Target distribution:")
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y,
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_features = [
    "alertscore",
    "episodealertscore",
    "longitude",
    "latitude",
    "year",
    "month",
    "duration_days_calculated",
]

categorical_features = ["alertlevel"]

numeric_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore")),
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features),
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Processed X_train shape:", X_train_processed.shape)
print("Processed X_test shape:", X_test_processed.shape)

assert X_train_processed.shape[0] == X_train.shape[0]
assert X_test_processed.shape[0] == X_test.shape[0]
assert not pd.isna(X_train_processed).any()
assert not pd.isna(X_test_processed).any()

print("Preprocessing completed successfully.")
```
