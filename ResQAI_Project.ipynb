```python
drought["disaster_type"] = "Drought"
earthquake["disaster_type"] = "Earthquake"
eruption["disaster_type"] = "Eruption"
flood["disaster_type"] = "Flood"
forest_fires["disaster_type"] = "Forest Fire"
tropical_cyclone["disaster_type"] = "Tropical Cyclone"
```

```python
def standardize_columns(df):
    df = df.copy()
    df.columns = df.columns.str.strip()

    rename_map = {
        "Description": "description",
        "Alertlevel": "alertlevel",
        "Alertscore": "alertscore",
        "Episodealertlevel": "episodealertlevel",
        "Episodealertscore": "episodealertscore",
        "Country": "country",
        "Fromdate": "fromdate",
        "Todate": "todate",
        "Iso3": "iso3",
        "Gdacs id": "gdacs_id",
        "GDACS ID": "gdacs_id",
        "GDACS_ID": "gdacs_id",
        "Longitude": "longitude",
        "Latitude": "latitude"
    }

    df = df.rename(columns=rename_map)

    return df
```

```python
drought = standardize_columns(drought)
earthquake = standardize_columns(earthquake)
eruption = standardize_columns(eruption)
flood = standardize_columns(flood)
forest_fires = standardize_columns(forest_fires)
tropical_cyclone = standardize_columns(tropical_cyclone)
```

```python
if "Unnamed: 0" in flood.columns:
    flood = flood.drop(columns=["Unnamed: 0"])

print(flood.columns.tolist())
```

```python
master_columns = [
    "description",
    "alertlevel",
    "alertscore",
    "episodealertlevel",
    "episodealertscore",
    "country",
    "fromdate",
    "todate",
    "iso3",
    "gdacs_id",
    "longitude",
    "latitude",
    "disaster_type"
]
```

```python
def prepare_for_merge(df):
    df = df.copy()

    for column in master_columns:
        if column not in df.columns:
            df[column] = np.nan

    return df[master_columns]
```

```python
drought = prepare_for_merge(drought)
earthquake = prepare_for_merge(earthquake)
eruption = prepare_for_merge(eruption)
flood = prepare_for_merge(flood)
forest_fires = prepare_for_merge(forest_fires)
tropical_cyclone = prepare_for_merge(tropical_cyclone)
```

```python
master_disaster_dataset = pd.concat(
    [
        drought,
        earthquake,
        eruption,
        flood,
        forest_fires,
        tropical_cyclone
    ],
    ignore_index=True
)
```

```python
master_disaster_dataset["year"] = master_disaster_dataset["fromdate"].dt.year

master_disaster_dataset["month"] = master_disaster_dataset["fromdate"].dt.month

master_disaster_dataset["duration_days"] = (
    master_disaster_dataset["todate"] -
    master_disaster_dataset["fromdate"]
).dt.total_seconds() / (24 * 60 * 60)

master_disaster_dataset["duration_days"] = master_disaster_dataset["duration_days"].round(2)

print("Date features created successfully.")
```
